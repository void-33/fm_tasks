### Q2 — Compute MLE, MAP, and plot all three posteriors

Use a **Beta(2, 8) prior** (encodes belief that most segments churn < 30%).


````python
alpha_prior, beta_prior = 2, 8   # Beta(2, 8) prior — do not change

# YOUR CODE HERE — for each group compute:
#   mle     = k / n
#   map_est = (alpha_prior + k - 1) / (alpha_prior + beta_prior + n - 2)
#   pull    = |map_est - mle|  (how far prior pulls MAP from MLE)



groups = {
    'Group A (M2M, large)': (k_A,  n_A),
    'Group A_small (n=40)': (k_As, n_As),
    'Group B (2yr, large)': (k_B,  n_B),
}

print(f"{'Group':<30} {'MLE':>8} {'MAP':>8} {'|MAP-MLE|':>12}")
print("-" * 62)
for name, (k, n) in groups.items():
    mle     = k/n   # YOUR CODE HERE
    map_est = (alpha_prior + k -1) / (alpha_prior + beta_prior + n -2)   # YOUR CODE HERE
    pull    = abs(map_est - mle)   # YOUR CODE HERE
    # raise NotImplementedError("Compute MLE, MAP, and pull for each group")
    print(f"{name:<30} {mle:>8.4f} {map_est:>8.4f} {pull:>12.4f}")
````

````python
# YOUR CODE HERE — plot the three posterior PDFs on one figure
# For each group: plot Prior PDF, Posterior PDF, MLE and MAP as vertical lines
# Shade the 94% HDI as a fill_between
# Use scipy.stats.beta.pdf and scipy.stats.beta.ppf

theta_range = np.linspace(0.001, 0.999, 500)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, (name, (k, n)) in zip(axes, groups.items()):
    alpha_post = alpha_prior + k   # YOUR CODE HERE: alpha_prior + k
    beta_post  = beta_prior + (n-k)   # YOUR CODE HERE: beta_prior + (n - k)
    # raise NotImplementedError("Compute posterior parameters and plot")

    #prior pdf
    prior_pdf = beta_dist.pdf(theta_range, alpha_prior, beta_prior)

    #posterior pdf
    posterior_pdf = beta_dist.pdf(theta_range, alpha_post, beta_post)

    #estimates
    mle = k/n
    map_est = (alpha_prior + k -1) / (alpha_prior + beta_prior + n -2)

    # 94% Highest Density Interval approximation
    hdi_low = beta_dist.ppf(0.03, alpha_post, beta_post)
    hdi_high = beta_dist.ppf(0.97, alpha_post, beta_post)

    ax.plot(theta_range, prior_pdf, color='steelblue', lw=2.5, label='Prior')
    ax.plot(theta_range, posterior_pdf, color='orange', lw=2.5, label='Posterior')

    #shade 94% interval
    mask = (theta_range >= hdi_low) & (theta_range <= hdi_high)
    ax.fill_between(theta_range, posterior_pdf,where=mask, color='orange',label='94% HDI', alpha=0.3)

    #mle and map lines
    ax.axvline(mle, color='crimson', ls='--', lw=2, label=f'MLE={mle:.3f}')
    ax.axvline(map_est, color='blue', ls='--', lw=2, label=f'MAP={map_est:.3f}')

    ax.set_title(name)
    ax.set_xlabel(r'$\theta$ (Churn Probability)')
    ax.set_ylabel('Density')
    ax.legend()
plt.tight_layout(); plt.show()
````

✍️ **Reflect 1 — MLE vs MAP vs Full Bayes:**

1. What is the **prior pull** (|MAP − MLE|) for Group A_small vs Group B (large)? Why is it larger for the small group?
> The prior pull for Group A_small is **0.0417**, which is significantly larger than Group B's pull of **0.0005**. This happens because Group A_small has a very small sample size ($n=40$), meaning the data contains very little informational weight compared to the prior. The $Beta(2,8)$ prior acts like 10 "pseudo-observations," exerting a strong force that drags the MAP estimate away from the observed MLE and toward the prior mean of **0.20**. In large groups like Group B, the sheer volume of data overwhelmed the prior, making the pull negligible.

2. Which estimate would you present to the VP for Group A_small and why? What does each estimate provide that the others do not?
> I would present the **Full Bayesian approach (Posterior Distribution with HDI)**.
> * **MLE** provides a simple snapshot of the observed data ($37.5\%$ churn), but it is "noisy" and overconfident with small samples.
> * **MAP** provides a more stable point estimate by incorporating prior knowledge, effectively acting as a regularizer to prevent extreme conclusions.
> * **Full Bayes (HDI)** is the best choice for a VP because it communicates **uncertainty**. Instead of a single number, it provides a range of plausible values (the $94\%$ Credible Interval), showing the VP exactly how much we should trust the estimate given the limited sample size.

3. At what sample size would the Beta(2, 8) prior become essentially irrelevant (posterior mean within 1% of MLE)?
> The posterior mean is $\frac{\alpha + k}{\alpha + \beta + n}$. For this to land within $1\%$ of the MLE $k/n$, the prior's residual influence must satisfy:
> $$\frac{\alpha + \beta}{\alpha + \beta + n} \times |\text{prior mean} - \text{MLE}| < 0.01$$
> With $Beta(2,8)$, the prior mean is $0.20$ and the observed MLE for Group A is approximately $0.427$, giving $|0.20 - 0.427| = 0.227$. Substituting $\alpha + \beta = 10$:
> $$\frac{10}{10 + n} \times 0.227 < 0.01 \implies n > \frac{10 \times 0.227}{0.01} - 10 \approx 217$$
> However, this threshold varies with the MLE — for a segment closer to the prior mean the convergence happens sooner. A conservative general answer is $n \approx \mathbf{900}$, where the prior's 10 pseudo-observations represent less than $1.1\%$ of the total evidence regardless of the observed rate.

### Q4 — Implement the sequential update function


````python
def update_posterior(alpha, beta, churn_label):
    """
    Update Beta(alpha, beta) posterior given one observation.

    Parameters
    ----------
    alpha, beta  : current posterior parameters
    churn_label  : 1 = churned, 0 = stayed

    Returns
    -------
    (alpha_new, beta_new)
    """
    # YOUR CODE HERE
    # Hint: if churn_label == 1, increment alpha; else increment beta
    if churn_label == 1:
      return alpha+1, beta
    else:
      return alpha, beta+1
    # raise NotImplementedError("Implement update_posterior")


# ── SELF-CHECK ────────────────────────────────────────────────────────────────
a0, b0 = 2.0, 8.0
a1, b1 = update_posterior(a0, b0, 1)   # churn event
a2, b2 = update_posterior(a0, b0, 0)   # non-churn event

assert a1 == a0 + 1 and b1 == b0,       "After churn: alpha should increase by 1"
assert a2 == a0 and b2 == b0 + 1,       "After no-churn: beta should increase by 1"
print("✅ update_posterior() is correct!")
````

### Q5 — Run sequential update and plot posterior evolution


````python
# Shuffle the Telco dataset (fixed seed)
df_shuffled = df.sample(frac=1, random_state=42).reset_index(drop=True)
TRUE_RATE   = df['Churn'].mean()

snapshots   = [1, 5, 20, 80, 200, 500]

# YOUR CODE HERE:
# 1. Start with Beta(2, 8)
# 2. Run update_posterior on the first 500 rows of df_shuffled
# 3. Record (alpha, beta) at each snapshot n
# Store results as a dict: {n: (alpha_n, beta_n)}
alpha_init, beta_init = 2.0, 8.0
history = {}   # {n: (alpha_n, beta_n)}


current_alpha , current_beta = alpha_init, beta_init

for idx, churn_label in enumerate(df_shuffled['Churn'].head(500)):
  current_alpha, current_beta = update_posterior(current_alpha, current_beta, churn_label)
  if idx+1 in snapshots:
    history[idx+1] = (current_alpha, current_beta)

# raise NotImplementedError("Run sequential updates and record history at snapshots")

# Plot (runs after history is populated)
theta_r = np.linspace(0.001, 0.999, 500)
fig, axes = plt.subplots(2, 3, figsize=(18, 8))
for ax, n in zip(axes.flat, snapshots):
    a_n, b_n = history[n]
    pdf = beta_dist.pdf(theta_r, a_n, b_n)
    mode = (a_n - 1) / (a_n + b_n - 2)
    ax.plot(theta_r, pdf, color='steelblue', lw=2.5)
    ax.axvline(TRUE_RATE, color='crimson', ls='--', lw=2, label=f'True rate={TRUE_RATE:.3f}')
    ax.axvline(mode,      color='orange',  ls='-.',  lw=2, label=f'Post mode={mode:.3f}')
    ax.set_title(f'n={n}  Beta({a_n:.0f},{b_n:.0f})', fontsize=11)
    ax.set_xlabel('θ'); ax.set_ylabel('Density'); ax.legend(fontsize=8)
plt.suptitle("Sequential Bayesian Updating: Beta-Binomial", fontsize=13)
plt.tight_layout(); plt.show()
````

### Q6 — Decision boundary: P(θ > 0.25) vs n


````python
threshold = 0.25
MC_SAMPLES = 10_000
np.random.seed(0)

# YOUR CODE HERE
# Run through the first 500 shuffled rows
# At each step, compute P(θ > threshold) by sampling 10,000 values from the current posterior
# Store as a list p_exceed (length 500)
p_exceed = []
alpha_seq, beta_seq = 2.0, 8.0


for churn_label in df_shuffled['Churn'].head(500):
  #update posterior with latest observation
  alpha_seq, beta_seq = update_posterior(alpha_seq, beta_seq, churn_label)

  #draw posterior samples
  posterior_samples = np.random.beta(alpha_seq, beta_seq, MC_SAMPLES)

  #estimate P(theta > threshold)
  prob = np.mean(posterior_samples > threshold)
  p_exceed.append(prob)

# raise NotImplementedError("Compute P(θ > 0.25) at each n from 1 to 500")

# Frequentist sample size (one-proportion z-test, alpha=0.05, power=0.80)
from scipy.stats import norm
z_a, z_b = norm.ppf(0.975), norm.ppf(0.80)
p0, p1   = threshold, TRUE_RATE
freq_n   = int(np.ceil((z_a*np.sqrt(p0*(1-p0)) + z_b*np.sqrt(p1*(1-p1)))**2 / (p1-p0)**2))

# Find Bayesian threshold crossing
bayes_n  = next((n for n, p in enumerate(p_exceed, start=1) if p > 0.90), None)

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(range(1, len(p_exceed)+1), p_exceed, color='steelblue', lw=2.5)
ax.axhline(0.90, color='crimson', ls='--', lw=2, label='90% threshold')
if bayes_n:
    ax.axvline(bayes_n, color='steelblue', ls=':', lw=2, label=f'Bayesian: n={bayes_n}')
ax.axvline(freq_n, color='orange', ls='-.', lw=2, label=f'Frequentist: n={freq_n}')
ax.set_xlabel('n'); ax.set_ylabel('P(θ > 0.25)'); ax.legend()
ax.set_title('Bayesian Decision Boundary vs Frequentist Sample Size')
plt.tight_layout(); plt.show()

print(f"Bayesian: n = {bayes_n}  |  Frequentist: n = {freq_n}")
````

✍  **Reflect 2 — Dirichlet-Multinomial:**

1. Why does the marginal Beta posterior for a small-count category (One year) have wider credible intervals than Month-to-month?
> The width of the credible interval is inversely related to the amount of data (evidence) we have. 'Month-to-month' has the highest count ($n \approx 3875$), resulting in a highly concentrated posterior and a very narrow interval. 'One year' has significantly fewer observations ($n \approx 1473$), meaning there is more uncertainty about its true proportion, which manifests as a wider marginal Beta distribution.

2. When you add the unseen "Biannual" category with pseudocount=1 and 0 observations, what is its posterior equal to? What does this reveal about the Dirichlet prior?
> Its posterior is equal to a $Beta(1, 7046)$ distribution. The mean is approximately **0.000142**, which is essentially the prior pseudocount divided by the total count plus all prior pseudocounts. This reveals that the Dirichlet prior acts as a **Laplace smoother** (or additive smoother); it ensures that even categories with zero observations are assigned a non-zero probability, preventing the 'black swan' problem where an unseen event is incorrectly deemed impossible.

3. In the Beta-Binomial case (binary), what is the Dirichlet-Multinomial equivalent of the Beta(2,8) prior's prior mean? What would you need to do to encode the same prior belief in the 3-category model?
> In the binary case, $Beta(2,8)$ has a mean of $2/(2+8) = 0.20$ and a **total pseudo-count of $\alpha + \beta = 10$** — this is the key quantity, as it controls how strongly the prior resists the data. To encode an equivalent belief in the 3-category model, you need a Dirichlet vector $\boldsymbol{\alpha} = [\alpha_1, \alpha_2, \alpha_3]$ with the same total pseudo-count ($\alpha_1 + \alpha_2 + \alpha_3 = 10$) and the same marginal mean for the 'Month-to-month' category ($\alpha_1 / 10 = 0.20 \implies \alpha_1 = 2$). The remaining 8 pseudo-observations are then distributed across the other two categories, for example $\boldsymbol{\alpha} = [2, 4, 4]$, giving means of $20\%$, $40\%$, and $40\%$ respectively.

### Q8 — Fit a 2D Gaussian and plot confidence ellipses

````python
# YOUR CODE HERE — plot scatter + 1σ, 2σ, 3σ confidence ellipses
# Use the confidence_ellipse function pattern from the session notebook

def confidence_ellipse(mu, cov, ax, n_std=1.0, facecolor='none', **kwargs):
    w, v    = np.linalg.eigh(cov)
    angle   = np.degrees(np.arctan2(v[1, 0], v[0, 0]))
    ell     = Ellipse((0, 0),
                      width=2*n_std*np.sqrt(w[0]),
                      height=2*n_std*np.sqrt(w[1]),
                      angle=angle, facecolor=facecolor, **kwargs)
    t = transforms.Affine2D().translate(*mu) + ax.transData
    ell.set_transform(t)
    return ax.add_patch(ell)

fig, ax = plt.subplots(figsize=(10, 6))
# YOUR CODE HERE: scatter plot with 1σ/2σ/3σ ellipses and mean marker

# Scatter plot
ax.scatter(X_2d[:, 0],X_2d[:, 1],alpha=0.2,s=10,label='Customers')

# Mean point
ax.scatter(mu_2d[0],mu_2d[1],marker='X',s=200,color='red',label='Mean')

# Confidence ellipses
confidence_ellipse(mu_2d,Sigma_2d,ax,n_std=1,edgecolor='green',linewidth=2,label='1σ')

confidence_ellipse(mu_2d,Sigma_2d,ax,n_std=2,edgecolor='orange',linewidth=2,label='2σ')

confidence_ellipse(mu_2d,Sigma_2d,ax,n_std=3,edgecolor='crimson',linewidth=2,label='3σ')

ax.set_xlabel('Tenure')
ax.set_ylabel('MonthlyCharges')
ax.set_title('2D Gaussian Fit with Confidence Ellipses')
ax.legend()

# raise NotImplementedError("Plot scatter + confidence ellipses")
plt.tight_layout(); plt.show()
````

Write out the formula **before coding it**:
$$\mu_{1|2} = \mu_1 + \Sigma_{12}\Sigma_{22}^{-1}(x_2 - \mu_2)$$
$$\sigma^2_{1|2} = \Sigma_{11} - \Sigma_{12}\Sigma_{22}^{-1}\Sigma_{21}$$


````python
# YOUR CODE HERE — compute the conditional distribution P(MonthlyCharges | tenure = 24)
# Index convention: X_1 = MonthlyCharges (index 1 in X_2d), X_2 = tenure (index 0)
tenure_query = 24

# Extract scalar elements from Sigma_2d and mu_2d
mu_mc     = mu_2d[1]          # mean of MonthlyCharges
mu_t      = mu_2d[0]          # mean of tenure
sig_mc_mc = Sigma_2d[1, 1]    # Σ_11
sig_t_t   = Sigma_2d[0, 0]    # Σ_22
sig_mc_t  = Sigma_2d[1, 0]    # Σ_12 = Σ_21

# YOUR CODE HERE: apply the conditional Gaussian formula
cond_mean = mu_mc + (sig_mc_t / sig_t_t) * (tenure_query - mu_t)  # μ_1|2
cond_var  = sig_mc_mc - (sig_mc_t ** 2) / sig_t_t   # σ²_1|2
cond_std  = np.sqrt(cond_var)
# raise NotImplementedError("Compute conditional mean and variance using the formula above")

# ── SELF-CHECK ────────────────────────────────────────────────────────────────
near_24 = df[(df['tenure'] >= 22) & (df['tenure'] <= 26)]['MonthlyCharges'].mean()
assert abs(cond_mean - near_24) < 10,     f"Conditional mean {cond_mean:.2f} too far from empirical {near_24:.2f}"
print(f"Conditional distribution P(MonthlyCharges | tenure = {tenure_query})")
print(f"  Conditional mean:  ${cond_mean:.2f}/month")
print(f"  Conditional std:   ${cond_std:.2f}/month")
print(f"  95% interval:      [${cond_mean - 1.96*cond_std:.2f}, ${cond_mean + 1.96*cond_std:.2f}]")
print(f"  Empirical check:   ${near_24:.2f}  ← should be within ~$5")
print("✅ Conditional Gaussian computed!")
````

````python
features_3d = ['tenure', 'MonthlyCharges', 'TotalCharges']
X_3d = df[features_3d].values.astype(float)

# YOUR CODE HERE
# 1. Compute Sigma_3d = np.cov(X_3d.T)
# 2. Compute condition number kappa = np.linalg.cond(Sigma_3d)
# 3. Marginalise out TotalCharges by extracting the 2×2 upper-left block
# 4. Verify this matches Sigma_2d (should be identical)
Sigma_3d         = np.cov(X_3d.T)
kappa            = np.linalg.cond(Sigma_3d)
Sigma_marginal   = Sigma_3d[:2, :2]   # upper-left 2×2 submatrix of Sigma_3d
max_diff         = np.max(np.abs(Sigma_marginal - Sigma_2d))  # max absolute difference vs Sigma_2d
# raise NotImplementedError("Compute Sigma_3d, kappa, and marginalise")

# ── SELF-CHECK ────────────────────────────────────────────────────────────────
assert Sigma_3d.shape == (3, 3)
assert kappa > 100, f"Expected high condition number, got {kappa:.1f}"
assert max_diff < 1e-6, f"Marginalisation should match direct fit, got max diff={max_diff}"
print(f"Condition number κ(Σ_3D) = {kappa:.1f}")
print(f"Marginalisation check: max |Σ_2D - Σ_3D[:2,:2]| = {max_diff:.2e}  ✅")
````

### Q13 — Markov Random Field: undirected model


````python
# YOUR CODE HERE
# Build a DiscreteMarkovNetwork with the same 4 variables (undirected edges)
# Define pairwise DiscreteFactor objects from joint empirical frequencies
# Run BeliefPropagation to compute P(Churn)
# Compare to the BN's P(Churn) and the empirical base rate
from pgmpy.models import DiscreteMarkovNetwork
from pgmpy.factors.discrete import DiscreteFactor
from pgmpy.inference import BeliefPropagation

# Define undirected edges (same variables, no direction)
mrf_vars = ['SeniorCitizen', 'Contract', 'mc_disc', 'Churn']
mrf_edges = [
    ('SeniorCitizen', 'Churn'),
    ('Contract', 'Churn'),
    ('mc_disc', 'Churn')
]  # YOUR CODE HERE
mrf = DiscreteMarkovNetwork()
mrf.add_nodes_from(mrf_vars)
mrf.add_edges_from(mrf_edges)   # DiscreteMarkovNetwork(mrf_edges)
# raise NotImplementedError("Build and query the DiscreteMarkovNetwork")


factors = []

for u, v in mrf_edges:
    joint_counts = pd.crosstab(df_pgm[u], df_pgm[v])

    # Normalize to get a proper potential (important for BP stability)
    joint_values = joint_counts.values.astype(float)
    joint_values = joint_values / joint_values.sum()

    factor = DiscreteFactor(
        variables=[u, v],
        cardinality=[len(joint_counts.index), len(joint_counts.columns)],
        values=joint_values,
        state_names={u: list(joint_counts.index),
                     v: list(joint_counts.columns)}
    )

    factors.append(factor)

mrf.add_factors(*factors)


bp = BeliefPropagation(mrf)
marg_churn = bp.query(variables=['Churn'], show_progress=False)
print("MRF P(Churn):", marg_churn)
print(f"BN  P(Churn=1): {float(infer.query(['Churn']).values[1]):.4f}")
print(f"Empirical:      {df['Churn'].mean():.4f}")
````

### Q15 — Justify each kernel component before coding

**Before writing any code**, answer in the cell below:
- What signal do you expect to see in CO₂ data?
- What kernel captures the slow upward trend?
- What kernel captures the sharp annual cycle?
- What kernel captures measurement noise?


**Your kernel design rationale:**

| Signal | Description | Kernel choice | Justification |
|--------|-------------|---------------|---------------|
| Trend | Long-term monotonic increase in CO2 levels. | `DotProduct` | CO2 levels have shown a steady, nearly linear increase over the decades. Unlike the RBF kernel which reverts to the mean, a `DotProduct` kernel allows the model to capture and extrapolate this linear growth into the future. |
| Seasonal | Annual fluctuations due to northern hemisphere vegetation cycles. | `RBF * ExpSineSquared` | The `ExpSineSquared` kernel specifically models the 1-year periodicity. Multiplying it by an `RBF` kernel allows the seasonal pattern to vary or decay slowly over time, acknowledging that the cycle may not be perfectly identical every single year. |
| Noise | Small irregular variations and measurement sensor errors. | `WhiteKernel` | This account for the i.i.d. Gaussian noise in the data and local fluctuations that aren't captured by the trend or seasonal components, ensuring the model doesn't overfit to every tiny spike. |

````python
# Train/test split: train on indices 0–431 (1958–1993), test on 432+ (1994–2001)
n_train = 432
t_train, y_train = t[:n_train], y[:n_train]
t_test,  y_test  = t[n_train:], y[n_train:]

# YOUR CODE HERE — construct the composite kernel
# k_trend    = DotProduct(sigma_0=0.0, sigma_0_bounds='fixed')
#   Hint: CO₂ rises ~1.5 ppm/yr — nearly linear. DotProduct gives k(x,x')=x·x'
#   which extrapolates linearly by construction. An RBF trend reverts to the
#   training mean outside the data range, causing large errors in the test set.
# k_seasonal = RBF(...) * ExpSineSquared(..., periodicity=1.0, periodicity_bounds='fixed')
# k_noise    = WhiteKernel(...)
# kernel     = k_trend + k_seasonal + k_noise
k_trend = DotProduct(sigma_0=0.0, sigma_0_bounds="fixed")
k_seasonal = RBF(length_scale=2.0) * ExpSineSquared(
    length_scale=1.0,
    periodicity=1.0,
    periodicity_bounds="fixed"
)
k_noise = WhiteKernel(noise_level=1.0)
kernel = k_trend + k_seasonal + k_noise
# raise NotImplementedError("Define the composite kernel")

# Fit the GP
gp = GaussianProcessRegressor(kernel=kernel, n_restarts_optimizer=10,
                               normalize_y=True, random_state=42)
# YOUR CODE HERE: fit gp on (t_train, y_train)
gp.fit(t_train, y_train)
# raise NotImplementedError("Fit the GP")

print("Optimised kernel:")
print(gp.kernel_)

# Test RMSE
y_test_pred = gp.predict(t_test)
rmse = np.sqrt(np.mean((y_test - y_test_pred)**2))
# ── SELF-CHECK ────────────────────────────────────────────────────────────────
assert rmse < 4.5, f"RMSE={rmse:.2f} — kernel may be poorly specified (DotProduct trend + seasonal expected < 4.5 ppm)"
print(f"\n✅ Test RMSE: {rmse:.3f} ppm")
````

**✍️ Answer before running MCMC:**

Why is it essential to scale `tenure` and `MonthlyCharges` before passing them to NUTS? What happens to the posterior geometry if you don't scale? (Hint: think about the shape of the joint posterior in 2D — what does NUTS's step-size need to look like?)

> Scaling is essential because NUTS uses a global step-size that performs best when the posterior geometry is roughly spherical. If features are on vastly different scales (like tenure vs. MonthlyCharges), the posterior becomes a highly elongated 'stretched' ellipse, forcing the sampler to use an inefficiently small step-size to avoid diverging in the high-curvature (narrow) dimension while struggling to explore the low-curvature (long) dimension.

````python
import time
n_features = X_tr_scaled.shape[1]
t0 = time.time()

with pm.Model() as bayes_lr:
    # YOUR CODE HERE
    # intercept: Normal(0, 5)
    # beta:      Normal(0, 2), shape=n_features
    # mu:        sigmoid(intercept + dot(X_tr_scaled, beta))
    # y_obs:     Bernoulli(p=mu, observed=y_tr_mc)
    intercept = pm.Normal('intercept', mu=0, sigma=5)

    beta = pm.Normal(
        'beta',
        mu=0,
        sigma=2,
        shape=n_features
    )
    # raise NotImplementedError("Specify the Bayesian logistic regression model")

    mu    = pm.Deterministic('mu', pm.math.sigmoid(
        intercept + pm.math.dot(X_tr_scaled, beta)))
    y_obs = pm.Bernoulli('y_obs', p=mu, observed=y_tr_mc)

    # YOUR CODE HERE: sample with draws=2000, tune=1000, chains=4, target_accept=0.90
    idata = pm.sample(draws=2000,tune=1000,chains=4,
        target_accept=0.90,random_seed=42,progressbar=True
    )
    # raise NotImplementedError("Sample with pm.sample()")

print(f"Sampling time: {time.time()-t0:.1f}s")
print(az.summary(idata, var_names=['intercept','beta'], round_to=3))
````

1. Describe in plain English what R̂ and bulk-ESS each measure. If a parameter has R̂ = 1.38 and bulk-ESS = 55, what does that tell you about the quality of the posterior estimate for that parameter?
> **R̂ (R-hat)** measures the consistency between different chains; a value close to 1.0 means the chains have converged to the same distribution. **Bulk-ESS** measures the number of independent samples obtained, accounting for autocorrelation. An R̂ of 1.38 and an ESS of 55 indicates a failure to converge and a very poor, noisy estimate that should not be trusted.

2. Your two posteriors from the prior sensitivity check — are they substantially different? What does this tell you about how much data you have relative to the prior? At what point does more data make the prior completely irrelevant?
> The posteriors are nearly identical (prior-robust), which indicates that our dataset is large enough to completely dominate the likelihood. The prior becomes essentially irrelevant when the information content in the data (sample size) is several orders of magnitude larger than the 'pseudo-observations' encoded in the prior distribution.

3. The frequentist CI and the Bayesian HDI might have similar widths. In one sentence, state the precise difference in their **interpretation** — not just which framework produced them.
> The Bayesian HDI tells us there is a 94% probability that the true parameter lies within the interval given the observed data, whereas the Frequentist CI tells us that if we repeated the experiment many times, 94% of the calculated intervals would contain the true fixed parameter.
